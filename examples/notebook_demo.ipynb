# Welcome to Pilot in Jupyter Notebooks!

Pilot is a self-hosted, zero-GC systems and application language built on a 14-token seed engine, 2-bit hardware type lattice, deterministic Perceus ARC, and sub-millisecond compilation.

This notebook demonstrates using canonical Pilot interactively with the **Pilot Jupyter Kernel**.

```pilot
println("Hello from Pilot interactive notebook!");
```

## 1. Defining Functions & State Continuity Across Cells

Declarations defined in one cell are seamlessly available in subsequent cells:

```pilot
fn fibonacci(n: i64) -> i64 {
    if n <= 1 {
        return n;
    }
    return fibonacci(n - 1) + fibonacci(n - 2);
}

let sequence_length: i64 = 10;
```

```pilot
let fib10 = fibonacci(sequence_length);
println(f"{fib10}");
```

## 2. Math & String Interpolation

```pilot
let pi: f64 = 3.14159;
let radius: f64 = 5.0;
let area: f64 = pi * radius * radius;

println(f"Radius: {5}");
println(f"Area: {area}");
```
