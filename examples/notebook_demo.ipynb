# Welcome to Pilot in Jupyter Notebooks!

Pilot is a high-performance modern programming language with zero-cost abstractions, real-time memory management (Perceus RC), and sub-millisecond compilation.

This notebook demonstrates using Pilot interactively with the **Pilot Jupyter Kernel**.

```pilot
println("Hello from Pilot interactive notebook!")
```

## 1. Defining Functions & State Continuity Across Cells

Declarations defined in one cell are seamlessly available in subsequent cells:

```pilot
fn fibonacci(n: int) -> int {
    if n <= 1 {
        return n
    }
    return fibonacci(n - 1) + fibonacci(n - 2)
}

let sequence_length = 10
```

```pilot
let fib10 = fibonacci(sequence_length)
fib10
```
